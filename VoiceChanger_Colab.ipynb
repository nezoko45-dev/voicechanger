```python
# 2) Clone/update the repo without deleting an existing checkout.
from pathlib import Path
repo = Path('/content/voicechanger')
if not (repo / '.git').exists():
    !git clone --depth 1 https://github.com/nezoko45-dev/voicechanger.git /content/voicechanger
else:
    %cd /content/voicechanger
    !git pull --ff-only
%cd /content/voicechanger
print('✅ Repo ready.')
```

```python
# 3) Install dependencies only if this runtime has not been prepared yet.
from pathlib import Path
marker = Path('/content/voicechanger/.colab_deps_installed')
if not marker.exists():
    !pip install -q -r requirements.txt
    marker.touch()
    print('✅ Dependencies installed.')
else:
    print('⚡ Dependencies already installed — skipping pip install.')
```
